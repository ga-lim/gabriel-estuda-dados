```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import tree
```

```python
df = pd.read_excel('../data/dados_cerveja.xlsx')
df
```

```python
target = df['classe']
target
```

```python
covariaveis = ['temperatura', 'copo', 'espuma', 'cor']
features = df[covariaveis]
features
```

```python
features['copo'] = features['copo'].map({'mud': 0, 'pint': 1})
features['espuma'] = features['espuma'].map({'não': 0, 'sim': 1})
features['cor'] = features['cor'].map({'clara': 0, 'escura': 1})

features
```

```python
model = tree.DecisionTreeClassifier()

model.fit(features, target)
```

```python
plt.figure(dpi=400)
tree.plot_tree(model,
               feature_names=covariaveis,
               class_names=model.classes_,
               filled=True)
```
