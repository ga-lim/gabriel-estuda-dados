```python
import pandas as pd
```

```python
idade = [31, 28, 40, 37, 35, 23, 43, 39, 22, 23, 30, 29, 41, 27, 32]

series_idades = pd.Series(idade)
series_idades
```

```python

media_idades = series_idades.mean()
summary_idades = series_idades.describe()

summary_idades
```

```python
# cada elemento dentro da Serie se associa a um índice, semelhante a uma chave de dicionário

print(series_idades[0]) # acessando elemento pelo seu índice

series_idades = series_idades.sort_values() # ordenando os valores da Serie
print(series_idades)

print(series_idades[0]) 
# ao tentar acessar novamente o elemento 0 após a ordenação, 
# o valor imprimido é o mesmo, já que a estrutura se comporta como chave:valor

print(series_idades.iloc[0]) # o iloc permite acessar o elemento a partir da posição dele dentro da Serie, que passa a se comportar como lista
```

```python
indexs = ["Mario", "Remi", "Ana", "Jonatas", "Olga", "Helena", "Ana", "Bia", 
          "Beto", "Pedro", "Ronaldo", "Vinicius", "Fernando", "Katia", "Carlos"]

# os índices das Series não precisam ser necessariamente númericos, nem únicos

series_idades = pd.Series(idade, index=indexs)
series_idades

series_idades['Ana'] # acessando o elemento no índice(chave) 'Ana', como possui mais de um, ele retornará todos que encontrar
```

Dataframes

```python
idade = [31, 28, 40, 37, 35, 23, 43, 39, 22, 23, 30, 29, 41, 27, 32]
nomes = ["Mario", "Remi", "Ana", "Jonatas", "Olga", "Helena", "Ana", "Bia", 
          "Beto", "Pedro", "Ronaldo", "Vinicius", "Fernando", "Katia", "Carlos"]

series_idades = pd.Series(idade)

df = pd.DataFrame() # cria um dataframe vazio

df["Idades"] = series_idades # adicionando uma Serie como uma nova coluna do dataframe e nomendo-a
df["Nomes"] = nomes # mesmo processo porém agora com uma lista
df
```

```python
print(df["Nomes"]) # Acessar uma coluna dentro do DataFrame a retorna como uma Serie

print(df.iloc[0]) # Acessa uma linha dentro do DataFrame e a retorna como uma Serie

# ao acessar um elemento dentro de um dataframe por coluna, o índice da Serie retornada será o mesmo do dataframe (0)
# ao acessar um elemento dentro de um dataframe por linha, o índice da Serie retornada será o nome das colunas (Idades, Nomes)
```

```python
# Ao acessar um elemento, seja por linha ou coluna, e obter uma Serie como resposta, é possível navegar dentro dela como uma Serie de fato

print(df.iloc[-1]["Nomes"]) # Acessando a linha do último elemento do DataFrame e obtendo seu nome
print(df.iloc[-1]["Idades"]) # Acessando a linha do último elemento do DataFrame e obtendo sua idade
```
