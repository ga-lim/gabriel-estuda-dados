```python
def amp_diff(x: pd.Series):
    amp = x.max() - x.min()
    media = x.mean()
    return np.sqrt((amp-media)**2)

def life_time(x: pd.Series):
    dt = pd.to_datetime(x)
    return (dt.max() - dt.min()).days
```

```python
idades = pd.Series([42, 23, 21, 30, 35, 40, 27, 12, 34, 60])
amp_diff(idades)
```

```python
df = transacoes.groupby(by=['IdCliente'], as_index=False).agg({
    'IdTransacao' : ['count'],
    'QtdePontos' : ['sum', 'mean', amp_diff],
    'DtCriacao' : [life_time]
})
df.head()
```
