Primeiro código de Aprendizado de Máquina: Árvore de Decisão

```python
import pandas as pd
from sklearn import tree
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('../data/dados_frutas.xlsx')
df
```

```python
model = tree.DecisionTreeClassifier(random_state=42)

# Target - Resposta
y = df['Fruta']
# Caracteristicas - Covariáveis
features = ['Arredondada', 'Suculenta', 'Vermelha', 'Doce']
X = df[features]
```

```python
y
```

```python
X
```

```python
# Aqui é o algoritmo de aprendizado, onde o modelo vai aprender os padrões e a máquina aprender como chegar a resposta correta.
model.fit(X, y)
```

```python
# Teste de Predição
model.predict([[1, 0, 0, 0]])
```

```python
plt.figure(dpi=400)
tree.plot_tree(model,
                feature_names=features,
                class_names=model.classes_,
                filled=True)
```

```python
proba = model.predict_proba([[1, 1, 1, 1]])[0]
pd.Series(proba, index=model.classes_)
```
