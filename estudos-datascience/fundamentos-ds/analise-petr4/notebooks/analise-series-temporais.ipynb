# Dados de séries temporais em painel

```python
df = pd.read_csv('../data/RRRP3.csv', index_col=0)
df['Date'] = pd.to_datetime(df.index)
df['Ativo'] = 'RRRP3'

df
```

```python
df2 = pd.read_csv('../data/PRIO3.csv', index_col=0)
df2['Date'] = pd.to_datetime(df2.index)
df2['Ativo'] = 'PRIO3'

df2
```

Definindo os dois df's como painel

```python
df_painel = df.set_index(['Ativo', 'Date']).copy()
df2_painel = df2.set_index(['Ativo', 'Date']).copy()
```

```python
df_painel
```

```python
df2_painel
```

```python
df_total = pd.concat([df_painel, df2_painel])
df_total = df_total.sort_index()

df_total
```

Retorno se trata da mudança percentual de uma variável em um período de tempo.

Se dá pela diferença entre a variável no tempo final pela variável no tempo inicial dividida pela variável no tempo final.

No python, o retorno pode ser calculado a partir da função .pct_change() do pandas.



**Interpretação do retorno**

Se o valor for positivo = Houve aumento

Se o valor for negativo = Houve queda

Se o valor for 0 = Não houve variação

```python
# Definindo o retorno do valor de fechamento das ações.

df_total['Retorno'] = df_total.groupby(level=0)['Close'].pct_change()

df_total
```

Estatísticas por ativo

```python
print('=== Estatísticas dos retornos por ativo ===')
df_total.groupby('Ativo')['Retorno'].describe().T
```

```python
# Volatilidade (Desvio Padrão)

print('=== Volatilidade dos retornos por ativo ===')
df_total.groupby('Ativo')['Retorno'].std()
```

Comparação temporal

```python
# Definindo uma tabela matricial (pivot) para fazer a comparação temporal dos valores de fechamento dos ativos.
pivot = df_total.reset_index().pivot(index='Date', columns='Ativo', values='Close')


print('\n=== Tabela pivot ===\n')
print(pivot.head())
```

Comparativo no Gráfico

```python
plt.figure(figsize=(10,5))
for col in pivot.columns:
    plt.plot(pivot.index, pivot[col], label=col)

plt.title('Comparativo de preços: RRR3 x PRIO3')
plt.xlabel('Data')
plt.ylabel('Preço')
plt.legend()
plt.grid(True)

plt.show()
```
