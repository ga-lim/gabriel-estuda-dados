Exercício - Quem teve mais transações de Streak

```python
import pandas as pd
```

```python
transacoes = pd.read_csv('../data/transacoes.csv', sep=';')
transacoes.head()
```

```python
transacoes_produto = pd.read_csv('../data/transacao_produto.csv', sep=';')
transacoes_produto.head()
```

```python
# 1;Airflow Lover;;lovers
# 10;Daily Loot;;rpg
# 11;Lista de presença;;present
# 12;Presença Streak;;present
# 13;R Lover;;lovers

produtos = pd.DataFrame({
    'IdProduto' : ['1', '10', '11', '12', '13'],
    'descProduto' : ['Airflow Lover', 'Daily Loot', 'Lista de presença', 'Presença Streak', 'R Lover'] 
})

produtos.head()
```

```python
cliente_transacao_produto = transacoes.merge(
right=transacoes_produto, 
how='left', 
on=['IdTransacao'])[['IdTransacao', 'IdCliente', 'IdProduto']]

cliente_transacao_produto
```

```python
df_full = cliente_transacao_produto.merge(
right=produtos, 
how='left', 
on=['IdProduto'])

df_full
```

```python
filtro = df_full['descProduto'] == 'Presença Streak'
streak = df_full[filtro].groupby(by=['IdCliente'])[['IdTransacao']].count()

streak.sort_values(by=['IdTransacao'], ascending=False).head(3)
```

Tentando fazer de maneira mais enxuta e performática

```python
produtos_filtro = produtos[produtos['descProduto'] == 'Presença Streak']

(transacoes.merge(right=transacoes_produto, how='left', on=['IdTransacao'])
            .merge(right=produtos_filtro, how='right', on=['IdProduto'])
            .groupby(by=['IdCliente'], as_index=False)[['IdTransacao']]
            .count()
            .sort_values(by=['IdTransacao'], ascending=False)
            .head(1)
)
```
